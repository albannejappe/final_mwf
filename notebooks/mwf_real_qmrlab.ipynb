```python
import scipy.io as sio
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Charger la vraie coupe extraite depuis qMRLab
mat = sio.loadmat("mwf_example_slice.mat")
t2w_real = mat["T2w_slice"]
mwf_real = mat["MWF_slice"]

fig_real = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Real Spin-Echo MRI (TE1)", "Real Quantitative MWF Map (%)")
)

fig_real.add_trace(
    go.Heatmap(
        z=np.flipud(t2w_real),
        colorscale="Gray",
        showscale=False,
        hovertemplate="Voxel (%{x}, %{y})<br>Signal: %{z:.0f}<extra></extra>"
    ),
    row=1, col=1
)

fig_real.add_trace(
    go.Heatmap(
        z=np.flipud(mwf_real),
        colorscale="Inferno",
        zmin=0, zmax=25,
        colorbar=dict(title="MWF (%)"),
        hovertemplate="Voxel (%{x}, %{y})<br>Real MWF: %{z:.1f}%<extra></extra>"
    ),
    row=1, col=2
)

fig_real.update_layout(
    title=dict(text="Real In Vivo Brain Slice (qMRLab Example Dataset)", x=0.5, xanchor="center"),
    template="plotly_white",
    height=480,
    width=820
)

fig_real.update_xaxes(showticklabels=False, row=1, col=1)
fig_real.update_yaxes(showticklabels=False, scaleanchor="x", scaleratio=1, row=1, col=1)
fig_real.update_xaxes(showticklabels=False, row=1, col=2)
fig_real.update_yaxes(showticklabels=False, scaleanchor="x2", scaleratio=1, row=1, col=2)

fig_real.show()
```
